```python
#Day 10
#Add 50 to the end.
#Insert 20 at index 2.
#Remove 25.
#Sort the list in ascending order.
#Reverse the list.
#Print the final list.

numbers = [10, 25, 5, 40, 15]

numbers.append(50)

numbers.insert(2,20)

numbers.remove(25)

numbers.sort(reverse = True)

#numbers.reverse()

print(numbers)


```

```python
#Find the largest number
#Find the smallest number
#Find how many times 7 appears
#Sort the list from smallest → largest
#Print the final list

numbers = [12, 45, 7, 23, 89, 34, 7, 56]

lnum = numbers[0]

for i in range(len(numbers)):
  if lnum < numbers[i]:
    lnum = numbers[i]

print(lnum)

snum = numbers[0]

for i in range(len(numbers)):
  if snum > numbers[i]:
    snum = numbers[i]

print(snum)

count = 0

for i in range(len(numbers)):
  if numbers[i] == 7:
    count = count + 1

print(count)

numbers.sort()

print(numbers)
```

```python
#Sets
#Students who know both Python and Java
#Students who know Python but NOT Java
#Students who know Java but NOT Python
#Students who know at least one of the two languages

python_students = {"Manish", "Rahul", "Arjun", "Kiran", "Vijay"}

java_students = {"Rahul", "Kiran", "Vijay", "Suresh"}

print(python_students & java_students)

print(python_students - java_students)

print(java_students - python_students)

print(python_students | java_students)
```

```python
#Dictionary
#Add:college → Manipal
#Change age from 24 to 25
#Change course from "CSE" to "AI & ML"

student = {
    "Name" : "Manish",
    "Age" : 24,
    "Course" : "CSE",
}

student["College"] = "Manipal"

student["Age"] = 25

student["Course"] = "AI&ML"

print(student)
```

```python
#pop

student = {
    "Name" : "Manish",
    "Age" : 24,
    "Course": "AI&ML",
    "College": "Manipal"
}

student.pop("Age")

student["Year"] = 1

student["Course"] = "CSE"

print(student)
```
