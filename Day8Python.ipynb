```python
#Day8
#Add 40 at the end
#Add 15 at index 2
#Remove 5
#Sort the list in ascending order
#Print the final list#

a = [20, 5, 30, 10, 25]

a.append(40)

a.insert(2,15)

a.remove(5)

a.sort()

print(a)
```

```python
#Sum of even numbers using for loop
a = [12, 5, 18, 7, 20, 3, 28, 11]
s = 0

for i in range(len(a)):
  if a[i] % 2 == 0:
    s = s + a[i]

print(s)
```

```python
#average of all positive numbers
a = [-5, 10, -2, 20, 15, -8, 30]
count = 0
s = 0

for i in range(len(a)):
  if a[i] > 0:
    s = s + a[i]
    count = count + 1

print(s/count)
```

```python
#Largest even number using for loop
a = [12, 5, 18, 7, 20, 3, 28, 11]
e = a[0]

for i in range(len(a)):
  if a[i]%2 == 0:
    if a[i] > e:
      e = a[i]

print(e)

```

```python
#Smallest odd number
a = [12, 5, 18, 7, 20, 3, 28, 11]

for i in range(len(a)):
    if a[i] % 2 == 1:
        odd = a[i]
        break

for i in range(len(a)):
    if a[i] % 2 == 1 and a[i] < odd:
        odd = a[i]

print(odd)
```

```python
#searching for a element
a = [10, 25, 7, 40, 15, 30]
target = 40

for i in range(len(a)):
  if a[i] == target:
    print('Found at index',i)
    break
else:
  print('Not found')
```

```python
#print duplicates
a = [10, 20, 30, 20, 20, 40, 10, 50]

for i in range(len(a)):
  for j in range(i+1, len(a)):
    if a[i] == a[j]:
      print(a[i])
      break
```

```python
#Find how many even numbers and how many odd numbers are present
a = [12, 5, 18, 7, 20, 3, 28, 11]

ecount = 0
ocount = 0

for i in range(len(a)):
  if a[i] % 2 == 0:
    ecount = ecount + 1
  else:
    ocount = ocount + 1

print(ecount)
print(ocount)
```

```python
#How many numbers are greater than 20
#How many numbers are divisible by 5
#Print both counts
a = [10, 25, 7, 40, 15, 30, 25, 18]

gcount = 0
dcount = 0

for i in range(len(a)):
  if a[i] > 20:
    gcount = gcount + 1
  if a[i] % 5 == 0:
    dcount = dcount + 1

print(gcount)
print(dcount)
```
