```python
#Day24
#pandas basics
# What is Pandas?
# Series
# DataFrame
# Creating data
# Columns
# Rows
# Indexing
# loc
# iloc
# Pandas is a Python library used for working with tabular data
```

```python
#A Series is basically one column of data.
import pandas as pd

marks = pd.Series([85,72,36,97])

print(marks)
```

```python
print(marks[0])
```

```python
#craeating a series with a custom index
import pandas as pd

marks = pd.Series([85,22,33],
                  index = ["Mani","Bhavi","Ani"])

print(marks)
```

```python
print(marks["Mani"])
```

```python
#DataFrame
#DataFrame is basically a table

date = {
    "Name" : ["Mani","Bhavi"],
    "Age" : [24,17],
    "Marks" : [33,66]
}

df = pd.DataFrame(date)

print(df)
```

```python
print(df["Name"])
```

```python
#Adding a new column
df["Passed"] = True

print(df)
```

```python
df["Bonus"] = df["Marks"]+5

print(df)
```

```python
# loc is used to select data based on labels/index names.
df.loc[0]
```

```python
df.loc[1]
```

```python
df.loc[1,"Name"]
```

```python
#iloc means integer-location based indexing
# his is VERY important for today.
# loc
# Uses labels:
# df.loc[1, "Marks"]
# iloc
# Uses positions:
# df.iloc[1, 2]

df.iloc[1]
```

```python
df.iloc[1,2]
```
