```python
#Day36

#Stratified Cross-Validation


# Cross-validation
#       ↓
# 5-fold CV
#       ↓
# Stratified K-Fold
#       ↓
# cross_val_score
#       ↓
# cross_validate
#       ↓
# multiple metrics
#       ↓
# macro / weighted / micro
#       ↓
# train vs test CV scores


# Stratified K-Fold tries to preserve the class proportions in every fold.
# So if the entire dataset is:
# 90% class 0
# 10% class 1
# each fold will try to maintain approximately:
# 90% class 0
# 10% class 1
# The key difference
# K-Fold:
# Split the data into K folds.
# Stratified K-Fold:
# Split the data into K folds while preserving the class distribution.
# That's the entire idea.
```

```python
# Next concept: shuffle=True
# Why do we often write:
# StratifiedKFold(
#     n_splits=5,
#     shuffle=True,
#     random_state=42
# )
# Suppose the dataset is arranged like:
# A A A A A A A A A A ... B B B B B
# If we don't shuffle, the arrangement of the original data can affect how folds are created.
# With:
# shuffle=True
# we randomly mix the samples before creating the folds, while stratification still tries to preserve the class ratio.



# And:
# random_state=42
# makes that random shuffle repeatable.
# So:
# Stratified = preserve class proportions
# Shuffle = mix the samples first
# Random state = make the mixing reproducible
```

```python
# Next: Macro vs Weighted vs Micro


# This is actually very important for your AML lab, so let's go slowly.
# Suppose we have 3 classes:
# Class A → 80 samples
# Class B → 15 samples
# Class C → 5 samples
# And the model gets:
# A precision = 0.90
# B precision = 0.80
# C precision = 0.40
# Now we have three ways to combine these scores.


# 1. Macro
# Treat every class equally:
# $$ Precision_{macro}=\frac{0.90+0.80+0.40}{3}=0.70 $$
# Even though Class C has only 5 samples, it gets the same importance as Class A.



# 2. Weighted
# Here, classes with more samples get more importance.
# Class A → 80 samples → high weight
# Class B → 15 samples → medium weight
# Class C → 5 samples  → low weight
# So Class A has a much bigger influence on the final score.
# Think:
# Weighted = classes are weighted according to their number of samples.
# This is useful when your dataset is imbalanced and you want the overall metric to reflect the actual class distribution.




# 3. Micro
# Micro goes a different route.
# Instead of calculating the metric separately for every class and then averaging, it combines the individual predictions across all classes first and calculates the metric globally.
# Think:
# Micro = treat all individual predictions together.
```

```python
# Requirements

# Load Iris.

# Create a RandomForestClassifier with:
# n_estimators=100
# random_state=42

# Use 5-fold Stratified Cross-Validation.

# Use cross_validate().
# Calculate:
# Accuracy
# Precision (macro)
# Recall (macro)
# F1 (macro)

# Return the training scores too.
# Print:
# individual test scores
# average test scores
# individual train scores
# average train scores



from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_validate

iris = load_iris()

X = iris.data
y = iris.target

model = RandomForestClassifier(n_estimators = 100, random_state = 42)

skf = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

results = cross_validate(model, X, y, cv = skf, scoring = ["accuracy", "precision_macro", "recall_macro", "f1_macro"], return_train_score=True)


#Individual test scores
print("Individual Test Accuracy:", results["test_accuracy"])
print("Individual Test Precision:", results["test_precision_macro"])
print("Individual Test Recall:", results["test_recall_macro"])
print("Individual Test F1:", results["test_f1_macro"])

#Average test score
print("\nAverage Test Accuracy:", results["test_accuracy"].mean())
print("Average Test Precision:", results["test_precision_macro"].mean())
print("Average Test Recall:", results["test_recall_macro"].mean())
print("Average Test F1:", results["test_f1_macro"].mean())


#Individual train scores
print("\nIndividual Train Accuracy :", results["train_accuracy"])
print("Individual Train Precision:", results["train_precision_macro"])
print("Individual Train Recall:", results["train_recall_macro"])
print("Individual Train F1:", results["train_f1_macro"])


#Average train scores

print("\nAverage train scores :", results["train_accuracy"].mean())
print("Average train Precision:", results["train_precision_macro"].mean())
print("Average train Recall:", results["train_recall_macro"].mean())
print("Average train F1:", results["train_f1_macro"].mean())
```
