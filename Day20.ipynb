```python
#Day20
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


d = Dog()

d.eat()
d.bark()
```

```python
class Animal:

    def sound(self):
        print("Animal sound")


class Dog(Animal):

    def sound(self):
        print("Bark")

a = Animal()
d = Dog()

a.sound()
d.sound()
```

```python
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

d = Dog("Bruno", "Labrador")

print(d.name)
print(d.breed)
```

```python
#Encapsulation
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

    def show_marks(self):
        print(self.__marks)


s = Student("Manish", 90)

print(s.name)
s.show_marks()
```

```python
#Iterators
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))
```

```python
def decorator(func):

    def wrapper():
        print("Before")
        func()
        print("After")

    return wrapper
@decorator
def greet():
    print("Hello")

greet()
```
