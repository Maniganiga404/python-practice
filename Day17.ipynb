```python
#ITERATORS AND GENERATORS

numbers = [10,20,30,40]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
```

```python
#Generators
def numbers():
    yield 10
    yield 20
    yield 30

gen = numbers()

print(next(gen))
```

```python
def even_numbers(n):
    for i in range(2, n + 1, 2):
        yield i

for num in even_numbers(10):
    print(num)
```

```python
# List	                                       Generator
# Stores all values immediately	               Generates values when needed
# Uses more memory for huge data	             Uses less memory
# Can access items repeatedly	                 Once consumed, values are gone
# Uses []                                      Uses () in generator expressions
```

```python
# Create a generator function called:
# even_numbers(n)
# It should generate all even numbers from 1 up to n

def even_numbers(n):
  for i in range(2, n+1,2):
    yield i

for num in even_numbers(10):
  print(num)
```
