```python
#Day26
#Pandas Data cleaning

# Missing values (NaN)
# Finding missing values — isna(), isnull()
# Removing missing values — dropna()
# Filling missing values — fillna()
# Removing duplicate rows — drop_duplicates()
# Detecting duplicates
# Fixing incorrect data
# Changing data types — astype()
# String cleaning — .str

import pandas as pd

df = pd.DataFrame({
    "Name": ["Manish", "Rahul", "Anu", "Kiran"],
    "Age": [20, None, 19, 22],
    "Marks": [85, 91, None, 88]
})

print(df)
```

```python
#Count missing values

print(df.isna().sum())
```

```python
# String Cleaning with .str

# " Manish " → extra spaces
# "RAHUL" → uppercase
# " anu " → extra spaces + lowercase

df = pd.DataFrame({
    "Name": [" Manish ", "RAHUL", " anu ", "Kiran"]
})

print(df)
```

```python
# Remove extra spaces — .str.strip()

df["Name"] = df["Name"].str.strip()
print(df)
```

```python
# Convert to lowercase — .str.lower()

df["Name"] = df["Name"].str.lower()
print(df)
```

```python
#Upper

df["Name"] = df["Name"].str.upper()
print(df)
```

```python
# Clean this DataFrame in the following order:
# 1. Clean the names
# Remove extra spaces and make all names lowercase.

# 2. Fix the incorrect age
# Replace -5 with 22.

# 3. Fill missing ages
# Fill missing Age values with 20.

# 4. Fill missing marks
# Fill missing Marks with 0.

# 5. Remove duplicate rows

# 6. Print the final cleaned DataFrame

import pandas as pd

df = pd.DataFrame({
    "Name": [" Manish ", "RAHUL", " Anu", "RAHUL", "Kiran "],
    "Age": [20, None, 19, None, -5],
    "Marks": [85, 91, None, 91, 88]
})

print(df)
```

```python
df["Name"] = df["Name"].str.strip()
df["Name"] = df["Name"].str.lower()
print(df)
```

```python
df["Age"] = df["Age"].replace(-5,22)
print(df)
```

```python
df["Age"] = df["Age"].fillna(20)
df["Marks"] = df["Marks"].fillna(0)
print(df)
```

```python
df.duplicated()
```

```python
df = df.drop_duplicates()
print(df)
```
