```python
name = "Python Programming"

v = ['a','e','i','o','u']

vcount = 0
ccount = 0

for ch in name:
  if ch in v:
    vcount = vcount + 1
  elif ch.isalpha():
    ccount = ccount + 1
print(vcount)
print(ccount)
```

```python
text = "Python123"
count = 0

for ch in text:
  if ch.isdigit():
    count = count+1
print(count)
```

```python
text = "Python 123"
lcount = 0
dcount = 0
scount = 0

for ch in text:
  if ch.isalpha():
    lcount = lcount + 1
  elif ch.isdigit():
    dcount = dcount + 1
  elif ch == " ":
    scount = scount + 1
print(lcount)
print(dcount)
print(scount)
```

```python
name = "madam"

rev = ""

for ch in name:
  rev = ch + rev
if name == rev:
  print("Palindrome")
else:
  print("Not a Palindrome")
```

```python
text = "Python123"

lcount = 0
dcount = 0

for ch in text:
  if ch.isalpha():
    lcount = lcount + 1
  elif ch.isdigit():
    dcount = dcount + 1

print(lcount)
print(dcount)
```

```python
#vowels,consonents,digits,spaces
text = "Python123 Programming"

v=['a','e','i','o','u']

vcount = 0
ccount = 0
dcount = 0
scount = 0

for ch in text:
  if ch in v:
    vcount = vcount+1
  elif ch.isalpha():
    ccount = ccount + 1
  elif ch.isdigit():
    dcount = dcount + 1
  elif ch == " ":
    scount = scount + 1
print(vcount)
print(ccount)
print(dcount)
print(scount)

```

```python
#Palindrome ignoring upper case
text = "Madam"

rev = ""

for ch in text:
  rev = ch + rev

if text.lower() == rev.lower():
  print("Palindrome")
else:
  print("Not a Palindrome")
```
