```python
#Day 9
#in a tuple find sum of even numbers
t = (12, 5, 18, 7, 20, 3, 28, 11)

esum = 0

for i in range(len(t)):
  if t[i] % 2 == 0:
    esum = esum + t[i]

print(esum)
```

```python
#Find the average of all positive numbers in this tuple
t = (-5, 10, -2, 20, 15, -8, 30)

pcount = 0
psum = 0

for i in range(len(t)):
  if t[i] > 0:
    pcount = pcount + 1
    psum = psum + t[i]

avg = psum / pcount
print(avg)
```
