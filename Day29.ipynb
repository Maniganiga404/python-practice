```python
# 3️⃣ What kind of ML models does scikit-learn provide?

# scikit-learn has tools for several types of ML:
# Regression → predict numbers
# Example: predict house price
# House features → ₹50 lakh

# Classification → predict categories
# Example: spam or not spam
# Email → Spam

# Clustering → group similar data
# Example: group customers based on behavior
# Customers → Group 1, Group 2, Group 3

# So you can think:
#               scikit-learn
#                    ↓
#         ┌──────────┼──────────┐
#         ↓          ↓          ↓
#    Regression  Classification Clustering
#         ↓          ↓          ↓
#      Numbers    Categories    Groups
```

```python
#from sklearn import neighbors
# We'll use K-Nearest Neighbors (KNN) because it's simple and it lets you see that the scikit-learn workflow stays almost the same.
# Our problem
# Suppose we have students:
# Hours Studied	Result
# 1	Fail
# 2	Fail
# 3	Pass
# 4	Pass
# 5	Pass

from sklearn.neighbors import KNeighborsClassifier

X = [[1],[2],[3],[4],[5]]
y = ["Fail","Fail","Pass","Pass","Pass"]

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X,y)

prediction = model.predict([[6]])

print(prediction)

```

```python
#Use KNN to predict whether a student will Pass or Fail based on study hours


from sklearn.neighbors import KNeighborsClassifier

X = [[1], [2], [3], [4], [5]]
y = ["Fail", "Fail", "Pass", "Pass", "Pass"]

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X,y)

predictions = model.predict([[6]])

print(predictions)
```
