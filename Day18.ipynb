```python
#Day18
# Create a decorator called log_function that prints:
# Function started
# then runs the original function, then prints:
# Function finished

def decorator(func):
  def log_functions(*args,**kwargs):
    print("function started")
    result = func(*args,**kwargs)
    print("function finished")
    return result
  return log_functions

@decorator
def greet(name):
  print("Hello",name)

greet("Manish")
```

```python
#Constructor
def multiplier(x):

    def multiply(y):
        return x * y

    return multiply


double = multiplier(2)
triple = multiplier(3)

print(double(10))
print(triple(10))
```

```python
# Create:
# @repeat(5)
# def hello():
#     print("Hello")
# so that:
# hello()
# prints:
# Hello
# Hello
# Hello
# Hello
# Hello

def repeat(n):
  def decorator(func):
    def wrapper(*args,**kwargs):
      for _ in range(n):
        func(*args,**kwargs)

    return wrapper

  return decorator

@repeat(5)
def hello():
  print("Hello")

hello()
```

```python
# timer
# def calculate():
#     # some work
# When we call:
# calculate()
# it should tell us approximately how long the function took

import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()

        print(f"Time taken: {end - start}")

        return result
    return wrapper


@timer
def calculate():
    total = 0

    for i in range(1000000):
        total += i

    return total



result = calculate()
print(result)
```
