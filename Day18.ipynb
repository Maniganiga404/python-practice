```python
#Day19
# Create a decorator called log_function that prints:
# Function started
# then runs the original function, then prints:
# Function finished

def decorator(func):
  def log_functions(*args,**kwargs):
    print("function started")
    result = func(*args,**kwargs)
    print("function finished")
    return result
  return log_functions

@decorator
def greet(name):
  print("Hello",name)

greet("Manish")
```

```python
class MyContext:

  def __enter__(self):
    print("Resource Opened")
    return self;

  def __exit__(self, exc_type, exc_value, traceback):
    print("Resource Closed")

with MyContext() as obj:
  print("Using resource")
```

```python
#will show us error but still perform cleanup
class MyContext:

    def __enter__(self):
        print("Starting")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Ending")
        print("Type:", exc_type)
        print("Value:", exc_value)

with MyContext():
    print(10 / 0)
```

```python
#creating exception handlers
class InvalidAgeError(Exception):
    pass

age = -5

if age < 0:
    raise InvalidAgeError("Age cannot be negative")
```

```python
try:
    age = -5

    if age < 0:
        raise InvalidAgeError("Age cannot be negative")

except InvalidAgeError as e:
    print(e)
```

```python
# Write a function:
# divide(a, b)
# Requirements:
# Try to divide a / b
# If b is 0, catch ZeroDivisionError
# Print "Cannot divide by zero"
# Use finally to print "Done"

def divide(a,b):
    try:
      result = a/b

    except ZeroDivisionError:
      print("Cannot divide by zero")

    finally:
      print("Done")

divide(10,0)
```

```python
# Create:
# class MyContext:
# It should:
# __enter__() print "Starting"
# __exit__() print "Ending"
# If an error happens inside the with block, print "Error occurred"
# If there is no error, just print "Ending"
# Test it with:
# with MyContext():
#     print("Working")

class MyContext:
    def __enter__(self):
        print("Starting")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            print("Error occurred")
        print("Ending")


with MyContext():
    print("Working")
```
