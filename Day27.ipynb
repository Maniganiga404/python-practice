```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

python_marks = [60, 70, 75, 85, 90]
java_marks = [50, 65, 70, 80, 85]

plt.plot(days, python_marks, label="Python")
plt.plot(days, java_marks, label="Java")

plt.title("Marks Progress")
plt.xlabel("Days")
plt.ylabel("Marks")

plt.legend()

plt.show()
```

```python
# Seaborn
# Seaborn is a Python library used to make better-looking statistical graphs more easily.
# First import it:
# import seaborn as sns


import seaborn as sns
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
marks = [40, 50, 55, 70, 80]

sns.scatterplot(x = hours, y = marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 55, 70, 80]
})

sns.scatterplot(data = df, x="Hours", y="Marks")

plt.show()
```

```python
#sns.lineplot()
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 55, 70, 80]
})

sns.lineplot(data = df, x = "Day", y = "Marks")

plt.title("Marks Progress")
plt.xlabel("Day")
plt.ylabel("Marks")

plt.show()
```

```python
#sns.barplot()

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Subject": ["Math", "Python", "Java", "DBMS"],
    "Marks": [85, 92, 78, 88]
})

sns.barplot(data=df, x="Subject", y="Marks")

plt.title("Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Marks")

plt.show()
```

```python
#sns.histplot()

import seaborn as sns
import matplotlib.pyplot as plt

marks = [35, 40, 42, 45, 50, 52, 55, 60, 65, 70,
         72, 75, 80, 85, 90]

sns.histplot(marks, bins=4)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
#sns.boxplot()

# A box plot is mainly useful for understanding:
# How the data is distributed
# The middle/typical value
# Spread of the data

import seaborn as sns
import matplotlib.pyplot as plt

marks = [35, 40, 42, 45, 50, 52, 55, 60, 65, 70, 72, 75, 80, 85, 95]

sns.boxplot(y=marks)

plt.title("Marks Distribution")
plt.show()
```

```python
df = pd.DataFrame({
    "Age": [18, 19, 20, 21, 22, 55]
})

sns.boxplot(data=df, y="Age")

plt.show()
```

```python
#sns.heatmap()

#A heatmap represents values using different shades/intensities, making patterns easy to spot

#In ML/EDA, one of the most common uses is showing a correlation matrix

import pandas as pd

import matplotlib.pyplot as plt

import seaborn as srs

df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Age": [18, 19, 20, 21, 22]
})

correlation = df.corr()

sns.heatmap(correlation, annot = True)

plt.title("Correlation Heatmap")

plt.show()

```

```python
#Visualization with Pandas

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Subject": ["Math", "Python", "Java", "DBMS"],
    "Marks": [85, 92, 78, 88]
})

df.plot(x = "Subject", y = "Marks", kind = "bar")

plt.show()
```

```python
# Question you're asking	Graph
# How does it change?	📈 Line
# Are these two related?	🔵 Scatter
# Which category is bigger?	📊 Bar
# How are values distributed + outliers?	📦 Box
# How are values distributed across ranges?	📊 Histogram
# How are many variables related?	🌡️ Heatmap
```

```python
#What is EDA?
#EDA = Exploratory Data Analysis

# EDA means looking at your data to understand what is happening inside it.
# Before building an ML model, we usually inspect the data.
# For example, we might ask:
# Are there missing values?
# What does the data look like?
# Are variables related?
# Are there unusual values/outliers?
# What patterns can we see?
# Visualization helps us answer these questions quickly


```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 150, 130, 180, 200],
    "Customers": [20, 25, 23, 30, 35]
})

# 1. Sales trend
sns.lineplot(data=df, x="Month", y="Sales")

# 2. Sales comparison
sns.barplot(data=df, x="Month", y="Sales")

# 3. Sales vs Customers
sns.scatterplot(data=df, x="Sales", y="Customers")

plt.show()
```
