```python
#Day35


# First, the big picture
# You already know:
# Random Forest
# Builds many decision trees → combines their predictions → majority vote.


# Cross-validation
# Uses different portions of the dataset to evaluate how well the model generalizes.
# So they solve different problems.


# 1. Why do we need Cross-Validation?
# Suppose we have 100 data points.
# Normally:
# Training = 80
# Testing  = 20
# You train on 80 and test on 20.
# But here's the problem:
# What if those particular 20 test samples are unusually easy?
# You might get:
# Accuracy = 95%
# But another set of 20 samples might give:
# Accuracy = 86%
# So we're asking:
# "Is my model actually good, or did I just get lucky with my train/test split?"
# Cross-validation helps answer that.
```

```python
# 2. 5-Fold Cross-Validation
# Let's say we have:
# 100 samples
# We divide them into 5 folds:
# Fold 1 → 20 samples
# Fold 2 → 20 samples
# Fold 3 → 20 samples
# Fold 4 → 20 samples
# Fold 5 → 20 samples


```

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

iris = load_iris()

X = iris.data
y = iris.target

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

scores = cross_val_score(model, X, y, cv=5)

print("Scores:", scores)
print("Average:", scores.mean())
```

```python
# Day 35 Practice Task
# Use the Iris dataset and create a Random Forest with:
# n_estimators = 100
# random_state = 42
# Then use:
# 5-fold cross-validation
# and calculate:
# The 5 individual accuracy scores
# The average accuracy

from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target

model = RandomForestClassifier(n_estimators=100, random_state=42)

scores = cross_val_score(model, X, y, cv=5, scoring = "accuracy")

print(scores)

print(scores.mean())
```
