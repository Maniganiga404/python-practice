```python
import pandas as pd

df = pd.DataFrame({
    "Name" : ["Manish", "Rahul", "Anu", "Kiran", "Priya"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 91, 76, 88, 95]
})

print(df)
```

```python
# 2. Selecting multiple columns

# to select one column.
# To select multiple columns, use a list:

print(df[["Name","Marks"]])
```

```python
# 3. Filtering rows ⭐
# This is one of the most important Pandas skills.
# Suppose we want students whose marks are greater than 85.

# First:
# df["Marks"] > 85
# creates a True/False result:
# 0    False
# 1     True
# 2    False
# 3     True
# 4     True
# Then:
# df[ ... ]
# keeps only the rows where the result is True.
# So:
# df[df["Marks"] > 85]

print(df[df["Marks"] > 85])
```

```python
print(df["Marks"]>85)
```

```python
print(df[df["Marks"] < 80])
```

```python
# 4. Combining conditions
# Suppose we want students who:
# are 20 or older AND scored more than 80.
# Use & for AND:

print(df[(df["Age"] >= 20) & (df["Marks"] > 80) ])
```

```python
#Marks greater than 90 OR age less than 20
print(df[(df["Marks"] > 90) | (df["Age"] < 20)])
```

```python
# Q1
# Print only Name and Age.
print(df[["Name", "Age"]])
```

```python
# Q2
# Print students whose marks are greater than 90
print(df[df["Marks"]>90])
```

```python
# Q3
# Print students whose age is less than 21
print(df[df["Age"]<21])
```

```python
# Q4
# Print students whose marks are greater than 80 AND age is 20 or more
print(df[(df["Marks"]>80) & (df["Age"]>=20)])
```

```python
#3. Adding a column

# We have:
# df

# Suppose we want to create a new column called Passed.
# df["Passed"] = df["Marks"] >= 80

# print(df)
# Pandas checks each row:
# 85 >= 80 → True
# 91 >= 80 → True
# 76 >= 80 → False
# 88 >= 80 → True
# 95 >= 80 → True

```

```python
#4. Modifying an existing column
df["Bonus"] = df["Marks"] + 10

print(df)
```

```python
#5. Removing a column

#Suppose we don't need Adult
# use:
# df = df.drop("Adult", axis=1)
# What does axis=1 mean?
# Think:
# axis=0 → rows
# axis=1 → columns

df = df.drop("Adult",axis = 1)

print(df)
```

```python
# Important thing about drop()
# This:
# df.drop("Adult", axis=1)
# does not normally change df itself.
# That's why we usually do:
# df = df.drop("Adult", axis=1)
# We're storing the changed DataFrame back into df.

```

```python
df["Marks"] = df["Marks"]+5

print(df)

df["Bonus"] = df["Marks"]+10

print(df)

df= df.drop("Bonus",axis=1)
print(df)

```

```python
#6. Sorting a DataFrame

#Sorting means arranging rows based on a column

import pandas as pd

df = pd.DataFrame({
    "Name": ["Manish", "Rahul", "Anu", "Kiran", "Priya"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 91, 76, 88, 95]
})

df = df.sort_values("Marks")

print(df)

#The index stays with the original row
```

```python
# Descending order
# To get largest → smallest:

df = df.sort_values("Marks", ascending = False)

print(df)
```

```python
#Sorting by multiple columns
# You can also sort using more than one column.

df = df.sort_values(["Marks", "Age"])

print(df)
```

```python
# One more useful method: head()
# Suppose you have 10,000 rows.
# You don't want to print everything.
# Use:
# print(df.head())
# It shows the first 5 rows by default.

print(df.head())
```
