```python
# 1️⃣ What is Logistic Regression?
# Despite its name, Logistic Regression is mainly used for classification.
# For example, suppose we want to predict:
# Student
#    ↓
# Pass or Fail
# The output has two classes:
# 0 → Fail
# 1 → Pass
# Or:
# Email
#    ↓
# Spam or Not Spam
# 0 → Not Spam
# 1 → Spam
# So Logistic Regression answers:
# "What is the probability that this sample belongs to class 1?"
# For example:
# Probability = 0.82
# The model can interpret that as:
# About 82% probability of class 1.
# Then we use a threshold to convert that probability into a class prediction.
# Probability
#     ↓
#   0.82
#     ↓
# Is it ≥ 0.5?
#     ↓
#    YES
#     ↓
# Class 1
# And if:
# Probability = 0.32
# then:
# 0.32 < 0.5
#       ↓
# Class 0
# The key idea
# Logistic Regression

# Features
#    ↓
# Calculate score
#    ↓
# Sigmoid function
#    ↓
# Probability (0 → 1)
#    ↓
# Threshold
#    ↓
# Class 0 or Class 1
```
