```python
#day16
#OOPs

# Create a class called BankAccount
# It should have:
# name
# balance
# Create a constructor using __init__()

class BankAccount:
    def __init__(self,name,balance):
      self.name = name
      self.balance = balance  #are attributes or data

    def deposit(self,amount):  #these are methods
      self.balance += amount

    def withdraw(self,amount):
      self.balance -= amount

    def display(self):
      print("Name:" ,self.name)
      print("Balance:" ,self.balance)
```

```python
#testing it

account1 = BankAccount("Manish",1000)

account1.deposit(500)

account1.withdraw(200)

account1.display()
```

```python
# Create a class called Student.
# Attributes:
# name
# marks
# Methods:
# display() → prints the student's name and marks.
# check_result() →
# If marks are 40 or above → print "Pass"
# Otherwise → print "Fail"


class Student:
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  def display(self):
    print("Name:",self.name)
    print("Marks:",self.marks)

  def check_result(self):
    if self.marks >= 40:
      print("Pass")
    else:
      print("Fail")
```

```python
s1 = Student("Manish",80)

s1.display()

s1.check_result()
```

```python
# area()
# perimeter()
# display()
# The formulas:
# Area = length × width
# Perimeter = 2 × (length + width)

class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width

  def area(self):
    area = self.length * self.width
    return area

  def perimeter(self):
    perimeter = 2 * (self.length + self.width)
    return perimeter

  def display(self):
    print("area:",self.area())
    print("perimeter:",self.perimeter())


```

```python
r1 = Rectangle(10,20)

r1.display()
```

```python
# Create a class called Product.
# Attributes
# name
# price
# quantity
# Methods
# total_price()
# Returns:
# price × quantity
# display()
# Prints:
# Product: <name>
# Price: <price>
# Quantity: <quantity>
# Total: <total_price>


class Product:
  def __init__(self,name,price,quantity):
    self.name = name
    self.price = price
    self.quantity = quantity

  def total_price(self):
    total_price = self.price * self.quantity
    return total_price

  def display(self):
    print("Product:",self.name)
    print("Price:",self.price)
    print("Quantity:",self.quantity)
    print("Total:",self.total_price())

```

```python
p1 = Product("Pen",300,3)

p1.display()
```

```python
#inheritance

# Practice: Person → Student
# Create a parent class called Person.
# Person should have:
# name
# age
# Use a constructor to initialize them.
# Then create a child class:
# class Student(Person):
# Student should additionally have:
# marks
# Use:
# super().__init__(name, age)
# Then create a method called display() that prints:

class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

class Student(Person):
  def __init__(self,name,age,marks):
    super().__init__(name,age)
    self.marks = marks

  def display(self):
    print("Name:",self.name)
    print("Age:",self.age)
    print("Marks:",self.marks)
```

```python
s1 = Student("Manish",25,90)

s1.display()
```

```python
#METHOD OVERRIDING

# Create a parent class Employee with a method:
# Then create a child class Manager(Employee).
# Inside Manager, override the work() method so it prints:
# Manager is managing the team

class Employee:
  def work(self):
    print("Employee is working")

class Manager(Employee):
  def work(self):
    print("Manager is managing the team")

m1 = Manager()
m1.work()
```

```python
#Polyorphism
# Create three classes:
# Shape
# Create a method:
# Rectangle
# Create an area() method that calculates:
# length × width
# Circle
# Create an area() method that calculates:
# π × radius × radius

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print("Calculating area:", self.length * self.width)


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Calculating area:", 3.14 * self.radius * self.radius)


r1 = Rectangle(10, 20)
c1 = Circle(10)

r1.area()
c1.area()
```

```python
class Shape:
    def area(self):
        print("Calculating area")


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

r1 = Rectangle(10, 20)
c1 = Circle(10)

print(r1.area())
print(c1.area())
```

```python
#create class
# Car
# class variable: wheels = 4
#  class method: change_wheels(cls, number)
#  Then call:
# Car.change_wheels(6)

# print(Car.wheels)

class Car:
  wheels = 4
  def __init__(self,brand):
    self.brand = brand

  @classmethod
  def change_wheels(cls,number):
    cls.wheels = number

  def display(self):
    print(self.brand)
    print(self.wheels)

Car.change_wheels(6)

c1 = Car("BMW")
c1.display()
```

```python
#staticmethod
# Create a class called Calculator with a static method:
# add(a, b)
# It should return the sum.
# Then call:
# print(Calculator.add(10, 20))

class Calculator:
  @staticmethod
  def add(a,b):
    total = a + b
    return total

print(Calculator.add(10,20))


```

```python
#Special Method
# Create:
# class Car:
# Attributes:
# brand
# model
# Add __str__() so that:
# c1 = Car("BMW", "M4")
# print(c1)

class Car:
  def __init__(self,brand, model):
    self.brand = brand
    self.model = model

  def __str__(self):
    return self.brand + " " + self.model

c1 = Car("BMW", "M4")

print(c1)
```
