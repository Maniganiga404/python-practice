```python
# min_samples_split
# Definition
# min_samples_split is the minimum number of samples a node must have before the Decision Tree is allowed to split it.
# For example:
# model = DecisionTreeClassifier(
#     min_samples_split=10,
#     random_state=42
# )



# min_samples_leaf
# This one is slightly different.
# Definition
# min_samples_leaf is the minimum number of samples that must remain in each leaf after a split.
# Example:
# model = DecisionTreeClassifier(
#     min_samples_leaf=5,
#     random_state=42
# )
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import plot_tree
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt

iris = load_iris()

X = iris.data
y = iris.target

model = DecisionTreeClassifier(
    min_samples_split=10,
    random_state=42
)

model.fit(X, y)

print(model.feature_importances_)

plt.figure(figsize=(15,10))


plot_tree(model,
          feature_names = iris.feature_names,
          class_names = iris.target_names,
          filled = True);

plt.show()
```
