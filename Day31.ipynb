```python
# What exactly is Regression?
# Regression is a type of supervised machine learning where a model learns
# the relationship between input features and a numerical target, so that it can predict a numerical value for new data
```

```python
# What is Linear Regression?
# Linear regression is a regression algorithm that tries to find the best straight-line relationship between the input and the numerical output.
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = [[2], [4], [6], [8], [10]]
y = [45, 55, 70, 82, 95]

model = LinearRegression()

model.fit(X,y)

predictions = model.predict([[20]])

print(predictions)

plt.scatter(X,y)

plt.plot(X, model.predict(X))

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

```python
# MAE — Mean Absolute Error
# Suppose the actual scores are:
# Actual:     50   60   70
# Predicted:  55   58   65
# Errors are:
# 5, 2, 5
# MAE simply takes the average of the absolute errors:
# $$ MAE = \frac{5+2+5}{3} = 4 $$
# So:
# MAE = average amount by which the model's prediction is wrong.
# And the key rule:
# Lower MAE = better model. ✅
```

<!-- MSE
Mean Squared Error is similar to MAE, but instead of just taking the absolute error, we square each error first.
For the same example:
Errors: 10, 20

20² = 400

MSE = (100 + 400) / 2
    = 250

```python
# MSE
# Mean Squared Error is similar to MAE, but instead of just taking the absolute error, we square each error first.
# For the same example:
# Errors: 10, 20

# Squared errors:
# 10² = 100
# 20² = 400

# MSE = (100 + 400) / 2
#     = 250
```

```python
# R² — R-squared
# R² tells us how well the regression model explains the variation in the target values.
# Think of it like a score:
# R² = 1 → model explains the data perfectly ✅
# R² = 0 → model isn't better than simply predicting the average
# Closer to 1 → generally better
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

actual = [50, 60, 70, 80, 90]
predicted = [52, 58, 73, 77, 88]

mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
r2 = r2_score(actual, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("R²:", r2)
```

```python
# Create a Linear Regression model that:
# Imports LinearRegression
# Creates the model
# Trains it using X and y
# Predicts the value for 6 hours
# Prints the prediction
# Calculates MAE, MSE, and R²

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X = [[1], [2], [3], [4], [5]]
y = [20, 30, 40, 50, 60]

model = LinearRegression()

model.fit(X,y)

predictions = model.predict([[6]])

print(predictions)

MAE = mean_absolute_error(y, model.predict(X))

MSE = mean_squared_error(y, model.predict(X))

R2 = r2_score(y, model.predict(X))

print(MAE)

print(MSE)

print(R2)
```
