```python
#find average of all numbers
a = [10, 25, 7, 40, 15, 30]
total = 0
n = len(a)

for i in range(len(a)):
  total = total + a[i]

avg = total/n
print(avg)
```

```python
#Largest even number
a = [12, 28, 18, 7, 20, 3]
m = a[0]

for i in range(len(a)):
  if a[i]%2==0:
    if a[i]>m:
      m=a[i]
print(m)
```
