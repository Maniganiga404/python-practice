```python
#Day12
#Compreshensions
numbers = [i*2 for i in range(1,11)]
print(numbers)
```

```python
#Create a list containing only the numbers divisible by 3 from 1 to 30
numbers = [i for i in range(1,31) if i%3 == 0]
print(numbers)
```

```python
#Create a list containing the squares of only the even numbers from 1 to 10
numbers = [i*i for i in range(1,11) if i % 2 == 0]
print(numbers)
```

```python
#Create a new list containing the names in uppercase using a list comprehension
names = ["manish", "rahul", "arjun", "kiran"]

names = [name.upper() for name in names]
print(names)
```

```python
names = ["manish", "rahul", "arjun", "kiran"]
length = [len(name) for name in names]
print(length)
```

```python

numbers = [i*i for i in range(1,6)]
print(numbers)
```

```python
#dictionary comprehensions
squares = {i: i*i for i in range(1,6)}
print(squares)
```

```python
#reate a dictionary containing only the odd numbers
numbers = {i : i * 10 for i in range(1,11) if i % 2 != 0}
print(numbers)
```

```python
#Create a dictionary where:
#key = name
#value = length of the name

name = {name: len(name) for name in names}
print(name)
```

```python
#Set comprehensions

numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6]

unique = {i*i for i in numbers if i% 2 == 0}
print(unique)
```
